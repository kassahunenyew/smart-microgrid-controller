```python
import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": 40.71,        
    "longitude": -74.01,      
    "start_date": "2023-01-01",
    "end_date": "2023-12-31",
    "hourly": ["shortwave_radiation", "temperature_2m", "cloudcover", "windspeed_10m"]
}

response = requests.get(url, params=params)
data = response.json()

df_solar = pd.DataFrame(data["hourly"])
df_solar["time"] = pd.to_datetime(df_solar["time"])
df_solar.set_index("time", inplace=True)
df_solar.to_csv("../data/solar_weather_data.csv")
print(df_solar.head())
print(f"\nShape: {df_solar.shape}")
```
